```python
## Batch Processing CTD Files
# KF 11/17/2020

# Setting up your python workspace
!pip install ctd
import ctd
import pandas as pd
import glob
```

```python
# Grab and sort all .cnv files in the directory
# Files have to be in the same directory as code 
files = glob.glob('*.cnv')
files = sorted(files)
files
```

```python
for datafile in files:
    cast = datafile.split('/')[-1].split('.')[0] # Extract cast from filename
    down, up = ctd.from_cnv(datafile).split()
    
    datadown = down.remove_above_water()\
           .despike()\
           .lp_filter()\
           .press_check()\
           .interpolate(method='index',
                        limit_direction='both',
                        limit_area='inside')\
           .bindata(delta=1, method='interpolate')\
           .smooth(window_len=21, window='hanning') 
    datadown = pd.DataFrame(datadown)
    datadown.index.name = 'PrDM'
    datadown.to_csv(cast + '_down.csv', index = True, header = True)
    
    dataup = up.remove_above_water()\
           .despike()\
           .lp_filter()\
           .press_check()\
           .interpolate(method='index',
                        limit_direction='both',
                        limit_area='inside')\
           .bindata(delta=1, method='interpolate')\
           .smooth(window_len=21, window='hanning') 
    dataup = pd.DataFrame(dataup)
    dataup.index.name = 'PrDM'
    dataup.to_csv(cast + '_up.csv', index = True, header = True)
```
